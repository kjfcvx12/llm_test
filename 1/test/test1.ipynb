```python
import pandas as pd

# 1. nsmc.py에서 찾은 원본 깃허브 주소 정의
_URL = "https://raw.githubusercontent.com/e9t/nsmc/master/"
_URLs = {
    "train": _URL + "ratings_train.txt",
    "test": _URL + "ratings_test.txt",
}

print("1. 네이버 영화 리뷰(NSMC) 테스트 데이터셋을 원본 URL에서 가져오는 중...")

# 2. test 주소를 사용하여 판다스로 직접 읽어오기 (탭 구분자 sep='\t')
df = pd.read_csv(_URLs["test"], sep='\t')

# 3. 데이터 안정성을 위해 내용이 비어있는 리뷰(결측치) 제거
df = df.dropna(subset=['document'])

df_sample=df.head(10).copy()

print(f"   => 성공적으로 {len(df_sample)}개의 전체 리뷰 데이터를 df로 가져왔습니다.\n")
print(df_sample.head())


# 2. 왓슨엑스 LLaMA 3.3 설정
model_id = "meta-llama/llama-3-3-70b-instruct"

parameters_greedy = {
    "decoding_method": "greedy",
    "max_new_tokens": 10  # '긍정' 또는 '부정' 단어만 받으므로 최소 토큰 설정
}

try:
    model = ModelInference(
        model_id=model_id,
        params=parameters_greedy,
        credentials=credentials,
        project_id=PROJECT_ID
    )
    
    # 3. LLaMA 3 프롬프트 빌더 함수
    def get_prompt(review_text):
        system_prompt = (
            "<|begin_of_text|><|start_header_id|>system<|end_header_id|>\n"
            "당신은 영화 리뷰 감성 분석기입니다. 입력된 한국어 리뷰를 분석하여 오직 '긍정' 또는 '부정' 중 하나의 단어로만 답변하세요. 다른 설명이나 문장은 절대 출력하지 마세요.<|eot_id|>"
        )
        few_shot = (
            "<|start_header_id|>user<|end_header_id|>\n리뷰: 진짜 최고의 영화입니다. 두 번 보세요!\n<|eot_id|>"
            "<|start_header_id|>assistant<|end_header_id|>\n긍정<|eot_id|>\n"
            "<|start_header_id|>user<|end_header_id|>\n리뷰: 돈 아깝고 시간 아까운 최악의 쓰레기 영화.\n<|eot_id|>"
            "<|start_header_id|>assistant<|end_header_id|>\n부정<|eot_id|>\n"
        )
        target = f"<|start_header_id|>user<|end_header_id|>\n리뷰: {review_text}\n<|eot_id|><|start_header_id|>assistant<|end_header_id|>\n"
        return system_prompt + few_shot + target

    # 4. 판다스 데이터프레임 순회하며 자동 분류 실행
    print("2. LLaMA 3.3 모델을 통한 자동 분류 시작...")
    print("=" * 60)
    
    predicted_labels = []
    
    for idx, row in df_sample.iterrows():
        review = row['document']
        actual_label = "긍정(1)" if row['label'] == 1 else "부정(0)"
        
        # 모델 호출
        prompt = get_prompt(review)
        response = model.generate_text(prompt=prompt)
        pred_result = response.strip()
        
        predicted_labels.append(pred_result)
        
        print(f"[{idx+1}/10] 리뷰: {review}")
        print(f"      👉 실제 정답: {actual_label} | 모델 예측: {pred_result}")
        print("-" * 60)

    # 5. 최종 결과를 데이터프레임에 추가 및 확인
    df_sample['model_prediction'] = predicted_labels
    print("\n3. 최종 결과 요약 데이터프레임:")
    print(df_sample[['document', 'label', 'model_prediction']])

except Exception as e:
    print("\n[작업 실패] 원인:", e)
```
